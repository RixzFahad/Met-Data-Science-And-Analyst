```python
import pandas as pd
```

```python
data = {
    'id': [1,2,3,4,5,6],
    'employee_name': ['Ser Duncan','Ser Arlan','Ser Steffon','Ser Raymun','Ser Duncan','Ser Arlan'],
    'department': ['Reach','Stormlands','Reach','Vale','Reach','Stormlands'],
    'sales': [1200,800,300,1500,1200,None],
    'profit': [200,150,50,300,200,120],
    'city': ['Delhi','Tokyo','London','Delhi','Delhi','Tokyo'],
    'email': ['a@mail.com','b@mail.com','c@mail.com','a@mail.com','a@mail.com','b@mail.com'],
    'date': ['2025-01-01','2025-01-02','2025-01-03','2025-01-04','2025-01-01','2025-01-02']
}

df = pd.DataFrame(data)
df
```

```python
#I See There Are Some Duplicate Data In This DataSet.
# Remove The Duplicate From Above DataSets
```

***Checking The Duplicates Values.***


```python
dupes = df[df.duplicated(subset=['employee_name', 'email'], keep=False)]
print(dupes)
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month_name()
df['day'] = df['date'].dt.day_name()
```

```python
df['sales'] = df['sales'].fillna(df['sales'].median())
df['department'] = df['department'].fillna('Missing')
```

```python
df
```

```python
import numpy as np

df['category'] = np.where(df['sales']>1000,'High',
                   np.where(df['sales']>=500,'Medium','Low'))
```

```python
df
```

```python
df['rolling_avg'] = df['sales'].rolling(3,min_periods=1).mean()
```

```python
df
```

```python
df.groupby(['department','date']).agg({'sales':['sum','mean','count']})
```

```python
df['city'] = df['city'].astype('category')
df.info(memory_usage='deep')
```
